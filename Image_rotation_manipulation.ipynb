```python
import PIL, os
from PIL import Image


def rotate_and_save(image_path):
    # Open an image file
    im = Image.open(image_path)
    rotated_img = im.rotate(90)
    rotated_img.show()

    # Save the rotated image to a file with suffix '_copy'
    base_name, e = os.path.splitext(image_path)
    png_path = base_name + "_copy.png"
    rotated_img.save(png_path, format='PNG')
    jpg_path = base_name + "_copy.jpg"
    rotated_img.save(jpg_path, format='JPEG')

rotate_and_save("katze.jpg")
```

```python

from PIL import Image, ImageEnhance

def manipulate_image(image_path):

    # Rotate the image 90 degrees clockwise
    im = Image.open(image_path)
    rotated_img = im.rotate(-90)
    rotated_img.show()
    
    # Convert to Grayscale
    grayscale_img = im.convert("L")
    grayscale_img.show()
    
    # Resize to 128x128
    resized_img = im.resize((128, 128))
    resized_img.show()

    # Increase brightness by 10%
    bright_enh = ImageEnhance.Brightness(im)
    bright_enh.enhance(1.1).show("10% more brightness")
    
    # Decrease Brightness by 10%
    bright_enh.enhance(0.9).show("reduced brightness by 10%")
    
    # Increase the contrast by 10%
    contrast_enh = ImageEnhance.Contrast(im)
    contrast_enh.enhance(1.1).show("increased contrast by 10%")
   
    # Decrease the contrast by 10%
    contrast_enh.enhance(0.9).show("reduced contrast by 10%")
    
manipulate_image("katze.jpg")
```

```python

from PIL import Image
import numpy as np

def image_manipulation_numpy(image_path):

    # Open an image file and convert to numpy array
    img = Image.open(image_path).convert("RGB")
    img_np = np.array(img)
    
    # Rotate the image 90 degrees clockwise
    rotated_np = np.rot90(img_np, k=3)
    rotated_img = Image.fromarray(rotated_np)
    rotated_img.show()
    
    # Convert to grayscale
    grayscale_np = np.dot(img_np[...,:3], [0.2989, 0.5870, 0.1140])
    grayscale_img = Image.fromarray(grayscale_np)
    grayscale_img.show()
    
   # Increase the brightness by 10%
    bright_inc = np.clip(img_np * 1.1, 0, 255).astype(np.uint8)
    bright_inc_img = Image.fromarray(bright_inc)
    bright_inc_img.show()

   # Decrease the brightness by 10%
    bright_dec = np.clip(img_np * 0.9, 0, 255).astype(np.uint8)
    bright_dec_img = Image.fromarray(bright_dec)
    bright_dec_img.show()

   # Increase the contrast by 10%
    alpha_inc = 1.1
    contrast_inc = np.clip(128 + alpha_inc * (img_np - 128), 0, 255).astype(np.uint8)
    contrast_inc_img = Image.fromarray(contrast_inc)
    contrast_inc_img.show(title="Contrast Increased")

    # Decrease the contrast by 10%
    contrast_dec = np.clip((img_np - 128) * 0.9 + 128, 0, 255).astype(np.uint8)
    contrast_dec_img = Image.fromarray(contrast_dec)
    contrast_dec_img.show()
    
image_manipulation_numpy("katze.jpg")
```
